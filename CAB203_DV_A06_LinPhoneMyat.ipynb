## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df = df.drop_duplicates()
df['Passengers'] = df['Passengers'].fillna(df['Passengers'].median())
df = df.dropna(subset=['Ticket_Type'])
df['Date'] = pd.to_datetime(df['Date'])
df.reset_index(drop=True, inplace=True)
````

### What all manipulations have you done and insights you found?

Duplicate values were checked and none were found. Missing values in 'Passengers' were filled using the median to maintain data distribution. Rows with missing 'Ticket_Type' were removed as it is important for analysis. The 'Date' column was converted to datetime format.

After cleaning, the dataset became consistent and suitable for analysis.

#### Chart - 14 - Correlation Heatmap

The extremely low correlation between Distance and Fare suggests a non-linear pricing model or the presence of flat-fare promotional trips (Tourist Cards), which warrants a review of the revenue leakage.
